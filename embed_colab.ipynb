# Embed Hansard chunks on a GPU
Runtime → Change runtime type → **T4 GPU**, then run the cell below and pick `data/index/texts.parquet` when asked. `dense.npy` downloads at the end; put it in `data/index/`.

```
from google.colab import files
import numpy as np, pandas as pd
from sentence_transformers import SentenceTransformer

files.upload()
texts = pd.read_parquet("texts.parquet").text.tolist()
model = SentenceTransformer("BAAI/bge-small-en-v1.5", device="cuda").half()
vectors = model.encode(texts, batch_size=256, normalize_embeddings=True, show_progress_bar=True)
np.save("dense.npy", vectors.astype(np.float16))
print(vectors.shape)
files.download("dense.npy")
```
